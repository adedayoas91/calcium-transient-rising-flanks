```python
from pathlib import Path
import json
import shlex
import subprocess
import sys

PACKAGE_ROOT = Path.cwd()
if not (PACKAGE_ROOT / 'pyproject.toml').exists():
    raise RuntimeError('Start this notebook from the package root')

RUN_OASIS = True
FLUO_TYPES = 'dff,f_smooth'
RECORDINGS = 'F3T1,F3T2,F5T2'
N_SURROGATES = 1000
OUTPUT_DIR = Path('outputs/revision_campaign/motorneurons_oasis')
command = [
    sys.executable, 'examples/empirical_baselines.py',
    '--components', 'oasis', '--fluo-types', FLUO_TYPES,
    '--recordings', RECORDINGS, '--oasis-outputs', 'spikes,denoised',
    '--cgc-methods', 'cgc,cgc-star',
    '--n-cgc-surrogates', str(N_SURROGATES),
    '--output-dir', str(OUTPUT_DIR), '--resume',
]
print(shlex.join(command))
if RUN_OASIS:
    subprocess.run(command, cwd=PACKAGE_ROOT, check=True)

summary_path = OUTPUT_DIR / 'summary.json'
if summary_path.exists():
    summary = json.loads(summary_path.read_text())
    print(json.dumps(summary, indent=2))
```
